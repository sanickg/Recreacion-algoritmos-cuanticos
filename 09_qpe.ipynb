## Estimación de fase (QPE)
Dado un operador U y uno de sus autovectores, estima la fase φ en U|ψ⟩ = e^{2πiφ}|ψ⟩.
Aquí U = compuerta de fase con φ = 1/8 (es la compuerta T). Es la base del algoritmo de Shor.

```python
import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Operator
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

sim = AerSimulator()

def correr(qc, shots=1024):
    """Ejecuta un circuito en el simulador y devuelve los conteos."""
    return sim.run(transpile(qc, sim), shots=shots).result().get_counts()
```

```python
def qpe(fase, n_cont):
    qc = QuantumCircuit(n_cont + 1, n_cont)
    qc.x(n_cont)                       # autovector |1> del operador de fase
    qc.h(range(n_cont))
    for k in range(n_cont):
        qc.cp(2*np.pi*fase * 2**k, k, n_cont)   # U^(2^k) controlada
    qc.append(qft(n_cont).inverse(), range(n_cont))
    qc.measure(range(n_cont), range(n_cont))
    return qc

fase, n_cont = 1/8, 3                  # prueba también fase=1/3 con n_cont=6 (no exacta)
conteos = correr(qpe(fase, n_cont))
mejor = max(conteos, key=conteos.get)
print("Fase real:", fase, "| Estimada:", int(mejor, 2) / 2**n_cont)
plot_histogram(conteos)
```
