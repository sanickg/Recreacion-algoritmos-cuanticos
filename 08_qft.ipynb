## Transformada cuántica de Fourier (QFT)
Es la versión cuántica de la transformada de Fourier discreta. Aquí la construimos a mano y
comprobamos que coincide con la matriz de la DFT.

```python
import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Operator
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

sim = AerSimulator()

def correr(qc, shots=1024):
    """Ejecuta un circuito en el simulador y devuelve los conteos."""
    return sim.run(transpile(qc, sim), shots=shots).result().get_counts()
```

```python
def qft(n, swaps=True):
    qc = QuantumCircuit(n, name="QFT")
    for j in reversed(range(n)):
        qc.h(j)
        for k in range(j):
            qc.cp(np.pi / 2**(j - k), k, j)
    if swaps:
        for i in range(n // 2):
            qc.swap(i, n - 1 - i)
    return qc

n = 3
N = 2**n
dft = np.array([[np.exp(2j*np.pi*j*k/N) for k in range(N)] for j in range(N)]) / np.sqrt(N)
print("¿Coincide con la DFT?", np.allclose(Operator(qft(n)).data, dft))
qft(4).draw("mpl")
```
